1. Using BPE from tiktoken

```python
from importlib.metadata import version

print("tiktoken version:", version("tiktoken"))
```

```python
import tiktoken

tik_tokenizer = tiktoken.get_encoding("gpt2")
text = "Hello, world. Is this-- a test?"
```

```python
integers = tik_tokenizer.encode(text)
print("Encoded integers:", integers)
```

```python
string = tik_tokenizer.decode(integers)
print("Decoded string:", string)
```

```python
print(tik_tokenizer.n_vocab)
```

2. Using the original BPE implementation used in GPT-2

```python
from bpe_openai_gpt2 import get_encoder, download_vocab
```

```python
download_vocab()
```

```python
orig_tokenizer = get_encoder(model_name="gpt2_model", models_dir=".")
```

```python
integers = orig_tokenizer.encode(text)
print("Encoded integers:", integers)
```

```python
string = orig_tokenizer.decode(integers)
print("Decoded string:", string)
```
